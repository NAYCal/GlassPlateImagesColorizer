# Overview

The goal of this assignment is to take the digitized Prokudin-Gorskii glass plate images and, using image processing techniques, automatically produce a color image with as few visual artifacts as possible. In order to do this, you will need to extract the three color channel images, place them on top of each other, and align them so that they form a single RGB color image. A cool explanation on how the Library of Congress created the color images on their site is available [here](http://www.loc.gov/exhibits/empire/making.html).

# Background

[Sergei Mikhailovich Prokudin-Gorskii](https://en.wikipedia.org/wiki/Sergey_Prokudin-Gorsky) (1863-1944) [Сергей Михайлович Прокудин-Горский, to his Russian friends] was a man well ahead of his time. Convinced, as early as 1907, that color photography was the wave of the future, he won Tzar's special permission to travel across the vast Russian Empire and take color photographs of everything he saw including the only color portrait of Leo Tolstoy. And he really photographed everything: people, buildings, landscapes, railroads, bridges... thousands of color pictures! His idea was simple: record three exposures of every scene onto a glass plate using a red, a green, and a blue filter. Never mind that there was no way to print color photographs until much later -- he envisioned special projectors to be installed in "multimedia" classrooms all across Russia where the children would be able to learn about their vast country. Alas, his plans never materialized: he left Russia in 1918, right after the revolution, never to return again. Luckily, his RGB glass plate negatives, capturing the last years of the Russian Empire, survived and were purchased in 1948 by the Library of Congress. The LoC has recently digitized the negatives and made them available on-line.

# Running the program
Running the program is relatively simple. First, run the cell with the libraries then go down any the following cells and click run. There will be explainations before each block of code to explain all the backend as intuitive as possible.

```python
# Important Libraries
import matplotlib.pyplot as plt

from models.default_images import DefaultImages
from models.colorize_glassplate_im import GlassPlateImage

ORIGINAL_IMAGES = [val.get_image() for val in DefaultImages]
ORIGINAL_TITLES = [val.value for val in DefaultImages]

def display_images(original_images, processed_images, title, original_titles=None, processed_titles=None):
    """
    Display original images and their corresponding processed images using matplotlib.

    :param original_images: List of original images.
    :param processed_images: List of processed images or list of list of images.
    :param title: The title of the entire plot.
    :param original_titles: List of titles for the original images.
    :param processed_titles: List of titles or list of list of titles for the processed images.
    """
    
    num_original = len(original_images)
    
    if all(isinstance(i, list) for i in processed_images):
        ncols = 1 + max(len(row) for row in processed_images)
    else:
        ncols = 2
        processed_images = [[img] for img in processed_images]

    fig, axes = plt.subplots(nrows=num_original, ncols=ncols, figsize=(5 * ncols, 5 * num_original))
    
    if num_original == 1:
        axes = axes.reshape(1, -1)

    for i in range(num_original):
        axes[i, 0].imshow(original_images[i], cmap='gray')
        if original_titles:
            axes[i, 0].set_title(original_titles[i])
        axes[i, 0].axis('off')
        
        for j, processed_img in enumerate(processed_images[i]):
            axes[i, j + 1].imshow(processed_img, cmap='gray')
            
            # Set the title for the processed image if provided
            if processed_titles:
                if isinstance(processed_titles[0], list):  # If it's a list of lists
                    axes[i, j + 1].set_title(processed_titles[i][j])
                else:  # If it's just a single list
                    axes[i, j + 1].set_title(processed_titles[j])
                
            axes[i, j + 1].axis('off')
    
    plt.suptitle(title, fontsize=16)
    plt.tight_layout()
    plt.show()
```
